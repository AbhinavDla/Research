# Run all autoencoder anomaly-cleaning experiments

Select the same Python kernel used by the experiment notebooks, then run the code cell below once. Set `SCENARIO_FILTER` to `"0-6"`, `"0-8"`, or `None` to run one scenario or both. Each selected scenario runs at `LATENT_DIM = 64` for `NUM_LATENT_SPACES = 5` and `10`.

Each run writes under `Iterations/<scenario>/<space count> Spaces/64 Latent Dimensions/<batch id>/`, with metrics in a text file and two image galleries displayed inline: 30 highest-scoring and 30 lowest-scoring images selected for removal during seed 1. Scores are relative suspicion ranks, not calibrated probabilities. The batch manifest is saved under `Iterations/Batch Runs/<batch id>/`.

```python
from pathlib import Path
import runpy

SCENARIO_FILTER = "0-6"  # Run only Train 0-6 / inject 7-9 for this comparison.

runner_name = "run_all_anomaly_cleaning_experiments.py"
search_roots = (Path.cwd().resolve(), *Path.cwd().resolve().parents)
runner_path = next(
    (candidate for root in search_roots
     for candidate in (root / runner_name,
                      root / "MNIST Latent Space Cleaning Verification" / "Autoencoder" / runner_name)
     if candidate.is_file()),
    None,
)
if runner_path is None:
    raise FileNotFoundError(f"Could not locate {runner_name} from {Path.cwd()}")

runpy.run_path(
    str(runner_path),
    init_globals={"SCENARIO_FILTER": SCENARIO_FILTER},
    run_name="__main__",
)
```
